```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(test_y, predictions)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Benign', 'Malignant'])
disp.plot(cmap='Blues')
plt.title('Confusion Matrix - Random Forest')
plt.show()
```

```python
#Out of 34 actual malignant cases in the test set, the model missed 5 (14.7% miss rate), while only 1 out of 52 benign cases was incorrectly flagged. In a medical screening context, false negatives are far more dangerous than false positives — a missed cancer diagnosis can delay life-saving treatment, whereas a false alarm just leads to a follow-up test. This means recall (sensitivity) is a more important metric here than overall accuracy, and future work should focus on reducing false negatives, even if it costs some precision.
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt
```

```python
probs = best_model.predict_proba(test_x)[:, 1] 
```

```python
fpr, tpr, thresholds = roc_curve(test_y, probs)
auc_score = roc_auc_score(test_y, probs)
```

```python
plt.figure(figsize=(6,6))
plt.plot(fpr, tpr, color='darkorange', label=f'ROC curve (AUC = {auc_score:.2f})')
plt.plot([0, 1], [0, 1], color='navy', linestyle='--', label='Random guess')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate (Recall)')
plt.title('ROC Curve - Random Forest')
plt.legend(loc='lower right')
plt.show()
```

```python
print(f'AUC Score: {auc_score:.3f}')
```
